**The daily driver.** Every cell launches a fresh process -- it can never be stale, never needs a kernel restart, and nothing here can overwrite the code.

Three rules:
1. New wire from Bloomberg -> paste it into `WIRE_TEXT` in the cell below, Shift+Enter.
2. Saved deal -> put its filename in `WIRE`, keep `WIRE_TEXT` empty, Shift+Enter.
3. Paste **dealer wires only** -- never a results table.
